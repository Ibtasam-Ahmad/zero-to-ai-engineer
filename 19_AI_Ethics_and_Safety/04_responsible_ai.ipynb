# Responsible AI

## Overview
Responsible AI covers the governance, documentation, safety, and regulatory frameworks needed to deploy AI systems that are transparent, accountable, and trustworthy. As AI becomes critical infrastructure, these practices move from optional to essential.

---

## 1. Core Principles

| Principle | Description |
|-----------|-------------|
| **Transparency** | Explain how the system works and what data it uses |
| **Explainability** | Users can understand individual decisions |
| **Accountability** | Clear ownership and redress mechanisms |
| **Fairness** | No discriminatory outcomes |
| **Privacy** | Data minimization, right to erasure |
| **Safety** | Robust, reliable, doesn't cause harm |
| **Human oversight** | Humans can intervene and correct |

---

## 2. EU AI Act (2024)

The world's first comprehensive AI regulation. Risk-based approach:

### Risk Categories

**Unacceptable Risk (Banned)**:
- Social scoring by governments
- Real-time biometric surveillance in public spaces (with narrow exceptions)
- Subliminal manipulation
- Exploiting vulnerabilities of specific groups

**High Risk (Heavy Regulation)**:
- Medical devices, safety systems
- Employment and HR decisions
- Education and vocational training
- Law enforcement, border control
- Credit scoring, insurance

Requirements: conformity assessment, human oversight, transparency, accuracy, robustness, data governance.

**Limited Risk (Transparency Obligations)**:
- Chatbots must disclose they are AI
- Deepfakes must be labeled

**Minimal Risk (No Regulation)**:
- Spam filters, AI in video games

### GPAI (General Purpose AI) Rules
- LLMs with >$10^{25}$ FLOPs training compute: additional requirements
- Systemic risk assessment
- Cybersecurity reporting

---

## 3. GDPR & Article 22

**Article 22**: Right not to be subject to solely automated decisions that produce legal or similarly significant effects.

Requirements when Article 22 applies:
- Meaningful information about the logic involved
- Right to human review
- Right to contest the decision

---

## 4. NIST AI Risk Management Framework (AI RMF)

Voluntary framework with 4 core functions:

1. **GOVERN**: Establish culture and accountability
2. **MAP**: Categorize AI risks in context
3. **MEASURE**: Analyze and assess risks
4. **MANAGE**: Prioritize and address risks

---

## 5. Model Documentation

### Model Cards (Google, Mitchell et al. 2019)
Structured documentation for ML models:
```
Model Details
Intended Use (primary, out-of-scope)
Factors (relevant demographic groups)
Metrics (performance metrics used)
Evaluation Data
Training Data
Quantitative Analyses (disaggregated evaluation)
Ethical Considerations
Caveats and Recommendations
```

### Datasheets for Datasets (Gebru et al. 2021)
Motivation, Composition, Collection Process, Preprocessing, Uses, Distribution, Maintenance.

---

## 6. LLM Safety

### Constitutional AI (Anthropic)
1. Generate harmful responses
2. Critique using a **constitution** (set of principles)
3. Revise based on critique
4. Train on revised outputs (RL-CAI)

### RLHF for Safety
- Human labelers rate outputs for harmlessness, helpfulness, honesty
- Reward model trained on ratings
- LLM fine-tuned with PPO to maximize reward

### Guardrails
- **Input guardrails**: classify/filter harmful inputs
- **System prompt instructions**: define model behavior and restrictions
- **Output guardrails**: check outputs before returning to user
- **NeMo Guardrails** (NVIDIA): programmable conversation flows
- **Guardrails AI**: output validation schemas

---

## 7. Red-Teaming LLMs

Structured adversarial testing:
- Manual red-teaming: human adversaries attempt to elicit harmful outputs
- Automated red-teaming: use another LLM to generate test prompts
- Red-teaming dimensions: bias, misinformation, harmful content, privacy leakage, jailbreaks

---

## 8. AI Watermarking & Deepfake Detection

**Watermarking LLM outputs**:
- Embed statistical signatures in generated text (specific token patterns)
- Green/red token lists: sample from green tokens with higher probability
- Detectable without the key being public

**Deepfake detection**:
- Artifact detection: GAN artifacts in images (spectral analysis)
- Temporal inconsistency in video deepfakes
- C2PA (Content Credentials): cryptographic provenance for images/video

---

## 9. Machine Unlearning

The challenge of removing specific training data influence from a model:
- **Exact unlearning**: full retraining without the data (expensive)
- **Approximate unlearning**: gradient ascent on forgotten data, fine-tuning
- Required for GDPR right to erasure compliance

````python
# ── Simple Guardrails Implementation ───────────────────────────
import re

BLOCKED_PATTERNS = [
    r'\b(bomb|explosive|weapon)\b',
    r'\b(how to hack|sql injection|xss attack)\b',
    r'\b(credit card number|ssn|social security)\b',
]

SENSITIVE_TOPICS = ['politics', 'religion', 'medical advice']

def input_guardrail(user_message: str) -> dict:
    """Screen input before sending to LLM."""
    message_lower = user_message.lower()
    
    # Check blocked patterns
    for pattern in BLOCKED_PATTERNS:
        if re.search(pattern, message_lower):
            return {'allowed': False, 'reason': f'Blocked pattern: {pattern}', 'sanitized': None}
    
    # Check length
    if len(user_message) > 10000:
        return {'allowed': False, 'reason': 'Message too long', 'sanitized': None}
    
    return {'allowed': True, 'reason': None, 'sanitized': user_message.strip()}

def output_guardrail(llm_response: str, max_length: int = 5000) -> dict:
    """Validate LLM output before returning to user."""
    if len(llm_response) > max_length:
        return {'safe': False, 'reason': 'Response too long', 'filtered': llm_response[:max_length]}
    
    for pattern in BLOCKED_PATTERNS:
        if re.search(pattern, llm_response.lower()):
            return {'safe': False, 'reason': 'Output contains blocked content', 'filtered': '[Response filtered]'}
    
    return {'safe': True, 'reason': None, 'filtered': llm_response}

# Test
test_inputs = [
    "What is machine learning?",
    "How do I make a bomb?",
    "Explain how sql injection works",
    "What are neural networks?"
]

print("Input Guardrail Tests:")
for msg in test_inputs:
    result = input_guardrail(msg)
    status = "✓ ALLOWED" if result['allowed'] else "✗ BLOCKED"
    print(f"  {status}: {msg[:50]}")
````

````python
# ── EU AI Act Risk Classification Helper ───────────────────────
def classify_ai_risk(use_case: str) -> str:
    """Simple heuristic for EU AI Act risk classification."""
    use_case_lower = use_case.lower()
    
    unacceptable = ['social score', 'real-time biometric', 'emotion recognition in school', 'subliminal']
    high_risk = ['hiring', 'recruitment', 'credit', 'medical', 'diagnosis', 'criminal',
                 'border control', 'critical infrastructure', 'education', 'insurance']
    limited_risk = ['chatbot', 'deepfake', 'synthetic']
    
    if any(term in use_case_lower for term in unacceptable):
        return "🔴 UNACCEPTABLE RISK Prohibited under EU AI Act"
    elif any(term in use_case_lower for term in high_risk):
        return "🟠 HIGH RISK Conformity assessment, human oversight, registration required"
    elif any(term in use_case_lower for term in limited_risk):
        return "🟡 LIMITED RISK Transparency obligations (must disclose AI)"
    else:
        return "🟢 MINIMAL RISK No specific obligations"

use_cases = [
    "AI for CV screening and hiring decisions",
    "Medical diagnosis assistance",
    "Customer service chatbot",
    "Spam filter for emails",
    "Real-time biometric surveillance in public spaces",
    "Credit scoring for loan applications"
]

print("EU AI Act Risk Classification:")
for uc in use_cases:
    risk = classify_ai_risk(uc)
    print(f"\n  Use case: {uc}")
    print(f"  Classification: {risk}")
````
