## 1. What is Jupyter?

**Jupyter** is a free tool that lets you write and run code in a web page called a **notebook** (a `.ipynb` file, like this one). A notebook mixes explanations, code, and the code's results in one document, which is why data scientists love it.

| Word | Meaning | Analogy |
|---|---|---|
| **Cell** | One box in the notebook. Either text (**Markdown**) or **code**. | One paragraph or one step of a recipe |
| **Kernel** | The Python program running in the background that actually executes your code and **remembers your variables**. | The cook in the kitchen; the notebook is just the recipe card |
| **Running a cell** | Sending that cell's code to the kernel. Shortcut: **Shift + Enter** (run and move down). | Asking the cook to do one step |
| **Execution count** | The number in `[ ]` to the left of a code cell, e.g. `[3]`. It shows the **order** cells were run in, not their position on the page. `[*]` means "still running". | A ticket number at the deli counter |
| **Output** | What appears below a code cell: printed text, tables, charts. | The finished dish |

### Handy keyboard shortcuts

Press **Esc** first to enter *command mode* (the cell border turns blue in classic Jupyter), or **Enter** to type inside a cell (*edit mode*).

| Shortcut | What it does |
|---|---|
| **Shift + Enter** | Run the cell, move to the next |
| **Ctrl + Enter** (Cmd + Enter on Mac) | Run the cell, stay on it |
| **Esc**, then **A** / **B** | Add a new cell **a**bove / **b**elow |
| **Esc**, then **M** / **Y** | Turn the cell into **M**arkdown / code |
| **Esc**, then **D D** | Delete the cell (press D twice) |
| **Esc**, then **0 0** | Restart the kernel (press 0 twice) |
| **Tab** | Auto-complete a name |
| **Shift + Tab** | Show help for the function under the cursor |

The exact shortcuts can differ a little between JupyterLab, classic Jupyter, VS Code and Google Colab; check the "Help" or "Keyboard shortcuts" menu in yours.

Let's run our first cell. Notice that the **last line's value is shown automatically**, even without `print`.

````python
print("Hello from the kernel!")   # print() always shows output
2 + 3                             # the LAST expression of a cell is displayed automatically
````

**What just happened?** The kernel ran both lines. `print(...)` showed its text, and because `2 + 3` was the **last line**, Jupyter displayed its value `5` as the cell's result (marked `Out[...]` in some Jupyter versions). Only the *last* line gets this treatment; a bare `2 + 3` in the middle of a cell shows nothing.

### Run order matters (the #1 notebook bug)

The kernel remembers variables in the **order you ran cells**, not the order they appear on the page. If you go back and re-run an earlier cell, later results can silently become out of date. Let's see it safely. Our demo variables all start with `demo_` so we can delete them afterwards.

````python
demo_price = 10                       # step A: set a price
demo_total = demo_price * 3           # step B: compute a total that DEPENDS on the price
print("price:", demo_price, "| total for 3 items:", demo_total)
````

**What just happened?** Nothing surprising yet: price 10, total 30.

Now imagine that, later, you **scroll up, change the price to 99, and re-run only that one line**. We simulate it in the next cell. The cell computing the total is *not* re-run.

````python
demo_price = 99                       # you changed step A and re-ran ONLY it...
print("price:", demo_price, "| total for 3 items:", demo_total)   # ...but step B was never re-run
print("What the total SHOULD be:", demo_price * 3)
````

**What just happened?** The price says 99 but the total still says **30**: it was computed from the *old* price and nothing recomputed it. In a long notebook this kind of stale value is very hard to spot, and it is the most common reason "it worked yesterday but not today".

⚠️ **Common mistake:** running cells out of order, deleting a cell that defined a variable, and the notebook still "working" because the kernel remembers the old value. The next person to run your notebook from the top gets an error.

### Restarting the kernel

**Restarting the kernel** wipes the kernel's memory: all variables and imports are forgotten, as if you just opened the notebook. In the menu: **Kernel → Restart** (or **Restart Kernel and Run All Cells** to rerun everything from the top). Doing *Restart and Run All* before you share a notebook is the best way to prove it really works in order.

We won't restart this notebook's kernel here, but we can **simulate** it for our demo variables with the magic command `%reset_selective` (more on magic commands soon), which deletes only variables whose names match a pattern.

````python
%reset_selective -f ^demo_
# -f = "don't ask for confirmation"; ^demo_ = "every name that starts with demo_"

try:
    print(demo_price)                 # this variable no longer exists in the kernel's memory
except NameError as err:
    print("NameError:", err)          # we CATCH the error so the notebook keeps running
````

**What just happened?** After the simulated restart, `demo_price` no longer exists, so Python raised a **`NameError`**: *"name 'demo_price' is not defined"*. That is exactly what you would see after a real restart if a cell used a variable defined in a cell you skipped or deleted.

We used `try:` / `except NameError:`: Python **tries** the code and, if that kind of error happens, runs the `except` part instead of stopping. You will learn this properly in [`01_basic.ipynb`](01_basic.ipynb); here we use it so the notebook can show errors without crashing.

💡 **Tip:** when a notebook acts strangely, the first fix to try is **Kernel → Restart and Run All**.

## 2. Markdown cells, help, tab completion and magic commands

### Markdown vs code cells

- A **code cell** contains Python. Running it sends the code to the kernel.
- A **Markdown cell** contains formatted text (like this one). "Running" it just displays the formatting. **Markdown** is a simple way to format text with symbols. Double-click any text cell in this notebook to see its Markdown.

| You type | You get |
|---|---|
| `# Big title`, `## Smaller title` | Headings |
| `**bold**`, `*italic*`, `` `code` `` | **bold**, *italic*, `code` |
| `- item` | A bullet point |
| `1. step` | A numbered list |
| `[link text](https://example.com)` | A clickable link |
| `| a | b |` lines | A table (like this one) |

### Getting help: `help()` and `?`

Every Python function has built-in documentation. Three ways to read it:
- `help(round)`: works in any Python, including scripts.
- `round?` in Jupyter: opens the help in a side panel or pop-up. `round??` also shows the source code when it is available.
- **Shift + Tab** with the cursor inside `round(`: a small help tooltip.

````python
help(round)          # prints the documentation of the built-in round() function
````

**What just happened?** `help` printed how to call `round`: `round(number, ndigits=None)`, meaning it takes a number and *optionally* how many digits to keep. Reading the help is often faster than searching the web.

### Tab completion

Type the first few letters of a name and press **Tab**: Jupyter shows the names that match. After a dot (for example `"hello".` then Tab) it shows everything that object can do. This saves typing and avoids typos. We can't press Tab inside a saved notebook, but the next cell does the same lookup by hand.

````python
text = "  Hello World  "
matches = [name for name in dir(text) if name.startswith("st")]   # names on a string that start with "st"
print("Typing  text.st  then Tab would offer:", matches)
print(text.strip())                                                # try one: removes the outer spaces
````

**What just happened?** `dir(text)` lists every method a string has; we kept those starting with `st`. Tab completion would show you `startswith` and `strip`. We then used `.strip()` to remove the spaces around `"Hello World"`.

### Magic commands

**Magic commands** are special Jupyter (IPython) commands that start with `%` (one line) or `%%` (the whole cell). They are **not Python**: they only work in notebooks and IPython, not in `.py` scripts.

| Magic | What it does |
|---|---|
| `%time <code>` | Runs the code **once** and reports how long it took |
| `%timeit <code>` | Runs the code **many times** and reports a reliable average |
| `%pwd` | Shows the current folder ("print working directory") |
| `%ls` | Lists files in a folder |
| `%pip install <package>` | Installs a package into the **same environment the kernel uses** |
| `%lsmagic` | Lists every magic command |

````python
%time total = sum(range(1_000_000))         # add up the numbers 0..999,999 once, and time it
%timeit -n 20 -r 3 sum(range(100_000))      # run 20 times, repeat 3 rounds, report the average
````

**What just happened?** `%time` reports the time of one run (look at "Wall time": the real elapsed time). `%timeit` ran the code 20 × 3 = 60 times and reported an average and spread. Your numbers depend on your computer: on a typical laptop the first takes some tens of milliseconds and the second a few milliseconds per loop. Use `%timeit` when you want to compare two ways of doing something; one run alone is noisy.

````python
from pathlib import Path        # pathlib: Python's modern way to work with file paths

here = %pwd
# %pwd returns the folder the kernel is working in. We print only the LAST part of it,
# because the full path contains your computer's user name and folder layout.
print("The kernel is working in a folder called:", Path(here).name)
````

**What just happened?** `%pwd` gave the **current working directory**, the folder the kernel treats as "here". Jupyter normally starts the kernel in the folder that contains the notebook, so you should see `01_Python`. That's why the notebooks in this repo can open files with short names like `notes/01_Python.md`.

⚠️ **Common mistake:** pasting full paths from your computer (like `C:\Users\yourname\...` or `/home/yourname/...`) into notebooks you share. They won't exist on anyone else's machine, and they reveal your user name.

Now `%ls`. To avoid listing your real files, we first make a **temporary folder** (a fresh, empty folder that we delete at the end) with Python's `tempfile` module, put two files in it, and list it.

````python
import tempfile, shutil, os, sys, subprocess    # tools for temp folders, files and running programs

workdir = Path(tempfile.mkdtemp(prefix="jupyter_demo_")).resolve()   # a brand-new empty folder
(workdir / "notes.txt").write_text("my notes\n")                      # create two small files in it
(workdir / "hello.py").write_text('print("Hello from a script!")\n')

def hide_tmp(text):
    "Replace the long temporary-folder path with <tmp> so outputs stay short and private."
    return str(text).replace(str(workdir), "<tmp>")

%ls $workdir
````

**What just happened?** `tempfile.mkdtemp()` created a new empty folder in your system's temporary area, and we wrote `notes.txt` and `hello.py` into it. `%ls $workdir` listed it (the `$` lets a magic command use a Python variable). On Linux/macOS a `*` after a name means "executable", a `/` means "folder". We also wrote a tiny helper, `hide_tmp`, that hides the long temporary path in later outputs.

### `%pip install`: explained, not run

To install a library from inside a notebook, use:

```
%pip install some-package-name
```

- `%pip` installs into **the same Python environment the kernel is using**, which is what you want.
- The older `!pip install ...` runs `pip` in a separate shell, which **may belong to a different Python** on your machine; the package then "installs fine" but `import` still fails.
- After installing, you may need to **restart the kernel** before the import works.

We don't run it here: this repo's libraries are installed once from `requirements.txt` (see [`START_HERE.md`](../START_HERE.md)), and installing things casually inside notebooks is how environments get messy.

## 3. The terminal in 10 minutes

The **terminal** (also called the *command line* or *console*) is a window where you type commands instead of clicking. The program that reads your commands is the **shell**: usually **bash** or **zsh** on Linux/macOS, and **PowerShell** (or the older **Command Prompt**, `cmd`) on Windows. You will use it to start Jupyter, install packages, run scripts and use Git.

Key ideas:
- **Current working directory (cwd)**: the folder the shell is "standing in". Commands act there unless you say otherwise.
- **Path**: the address of a file or folder.
  - An **absolute path** starts from the very top of the drive: `/home/ana/project/data.csv` (Linux/macOS) or `C:\Users\ana\project\data.csv` (Windows). Works from anywhere, but only on that one computer.
  - A **relative path** starts from the cwd: `data.csv`, `data/train.csv`, `../README.md` (`..` = "the folder above", `.` = "this folder"). Portable across computers, but depends on where you are.

### Command cheat sheet

| What you want | Linux / macOS (bash, zsh) | Windows PowerShell | Windows Command Prompt |
|---|---|---|---|
| Where am I? | `pwd` | `pwd` or `Get-Location` | `cd` |
| List files | `ls`, `ls -l` | `ls` or `dir` | `dir` |
| Go into a folder | `cd data` | `cd data` | `cd data` |
| Go up one folder | `cd ..` | `cd ..` | `cd ..` |
| Make a folder | `mkdir results` | `mkdir results` | `mkdir results` |
| Show a file | `cat notes.txt` | `cat notes.txt` or `type notes.txt` | `type notes.txt` |
| Copy / move / delete | `cp`, `mv`, `rm` | `cp`, `mv`, `rm` (aliases) | `copy`, `move`, `del` |
| Run a Python script | `python3 hello.py` (or `python`) | `python hello.py` (or `py hello.py`) | `python hello.py` |
| Clear the screen | `clear` | `cls` or `clear` | `cls` |
| Stop a running program | Ctrl + C | Ctrl + C | Ctrl + C |

⚠️ **Common mistake:** `rm` (and `del`) deletes files **permanently**: no recycle bin. Double-check before you press Enter.

### Trying it from the notebook

You can run terminal commands from Jupyter in two ways:
- Start a line with **`!`**, e.g. `!ls`: runs it in a shell (bash on Linux/macOS, cmd on Windows).
- Use Python's **`subprocess`** module: works in normal `.py` scripts too.

We only run commands **inside our temporary folder**. First, the `!` style, asking the Python the kernel uses for its version:

````python
!"{sys.executable}" --version
# sys.executable is the full path of the Python running this notebook; {braces} insert a Python value.
# In a terminal you would just type:  python --version
````

**What just happened?** The shell ran Python with the `--version` option and printed something like `Python 3.12.x` (this repo is tested on Python 3.12). Using `sys.executable` guarantees we ask *the same* Python that runs this notebook.

Now `subprocess`. The helper below runs a command **with the temporary folder as its cwd** and prints what it output. It picks the right command for your operating system (`os.name == "nt"` means Windows).

````python
WINDOWS = os.name == "nt"                         # True on Windows, False on Linux/macOS

def run(command, cwd=workdir):
    "Run a shell command inside `cwd` and print its output (with the temp path hidden)."
    result = subprocess.run(command, cwd=cwd, shell=isinstance(command, str),
                            capture_output=True, text=True)
    print(hide_tmp(result.stdout + result.stderr).rstrip())
    return result.returncode                       # 0 means success

run("cd" if WINDOWS else "pwd")                    # where am I?
run("dir /b" if WINDOWS else "ls")                 # list files
````

**What just happened?** `pwd` printed the cwd (shown as `<tmp>` because our helper hides the long path), and `ls` listed the two files we created. The number shown last is the **exit code** of `ls`: **`0` means success** (only the last line's value is displayed). Every terminal command returns one, and tools (like automated tests) use it to know whether something failed.

### Absolute vs relative paths, in Python

Python's `pathlib` makes paths work the same on every operating system (it uses `/` or `\` as needed).

````python
(workdir / "data").mkdir()                                   # make a sub-folder, like `mkdir data`
(workdir / "data" / "scores.csv").write_text("name,score\nana,90\n")

relative = Path("data") / "scores.csv"                       # RELATIVE: depends on the cwd
absolute = workdir / relative                                # ABSOLUTE: full address from the top
print("relative path :", relative)
print("absolute path :", hide_tmp(absolute))
print("is absolute?  :", relative.is_absolute(), "vs", absolute.is_absolute())
print("file exists?  :", absolute.exists())
print("relative path from the KERNEL's cwd exists?", relative.exists())   # we are not in workdir!
````

**What just happened?** Both paths point at the same file, but the **relative** one only works if you are standing in the right folder. The kernel's cwd is `01_Python`, not our temp folder, so `Path("data/scores.csv").exists()` is **False** there. This is the cause of most `FileNotFoundError`s (see section 5).

### Running a Python script

A **script** is just a `.py` text file of Python code. In a terminal you run it with `python hello.py`. Scripts can also receive **command-line arguments**: extra words after the script name, available in `sys.argv`.

````python
(workdir / "greet.py").write_text(
    "import sys\n"
    "name = sys.argv[1] if len(sys.argv) > 1 else 'stranger'   # first word after the script name\n"
    "print(f'Hello, {name}!')\n"
)
run([sys.executable, "hello.py"])               # like typing:  python hello.py
run([sys.executable, "greet.py", "Ana"])        # like typing:  python greet.py Ana
code_returned = run([sys.executable, "-c", "1/0"])   # a program that crashes
print("exit code of the crashing program:", code_returned)
````

**What just happened?**
- `hello.py` printed its message, just as if you had typed `python hello.py` in a terminal opened in that folder.
- `greet.py Ana` received `"Ana"` through `sys.argv[1]` and greeted you by name.
- The last program (`python -c "1/0"` runs one line of code) crashed with a **`ZeroDivisionError`** traceback and an exit code of **`1`** (not 0), which is how the terminal knows it failed. You will learn to read that traceback in section 5.

💡 **Tip:** if `python` is "not found", try `python3` (Linux/macOS) or `py` (Windows). Inside an activated virtual environment (next section), `python` always works.

## 4. Virtual environments, pip and requirements files

- **pip** is Python's package installer. It downloads libraries such as `numpy` from the Python Package Index (PyPI) and installs them.
- A **virtual environment** ("venv") is a **private folder of installed packages for one project**.

**Why bother?** Project A may need `numpy 1.x` while project B needs `numpy 2.x`. If everything is installed into one shared Python, upgrading for B can break A. With one venv per project, each has its own toolbox. It also means you can delete a broken environment and rebuild it in minutes.

### The commands (shown, not run)

| Step | Linux / macOS | Windows (PowerShell) |
|---|---|---|
| 1. Create a venv in a folder called `.venv` | `python3 -m venv .venv` | `py -m venv .venv` |
| 2. Activate it (your prompt then starts with `(.venv)`) | `source .venv/bin/activate` | `.venv\Scripts\Activate.ps1` |
| 3. Install a package | `python -m pip install numpy` | `python -m pip install numpy` |
| 4. Install everything a project needs | `python -m pip install -r requirements.txt` | same |
| 5. Save what you installed | `python -m pip freeze > requirements.txt` | same |
| 6. Make Jupyter see this venv as a kernel | `python -m ipykernel install --user --name myproject` | same |
| 7. Leave the venv | `deactivate` | `deactivate` |

Notes:
- `python -m pip` (instead of plain `pip`) guarantees you use the pip that belongs to the active Python.
- In Windows PowerShell, activation may first need: `Set-ExecutionPolicy -Scope CurrentUser RemoteSigned` (allows local scripts). In Command Prompt use `.venv\Scripts\activate.bat`.
- Never commit the `.venv` folder to Git; commit `requirements.txt` instead.
- Other tools (for example **conda** or **uv**) do the same job in different ways; the idea is identical.

This repo's exact setup steps are in [`START_HERE.md`](../START_HERE.md), and its package list is the root [`requirements.txt`](../requirements.txt).

Is *this* notebook running inside a virtual environment? Python can tell us:

````python
from importlib.metadata import version            # reads the version of an installed package

in_venv = sys.prefix != sys.base_prefix           # inside a venv these two locations differ
print("Running inside a virtual environment?", in_venv)
print("Python version:", sys.version.split()[0])
print("numpy version installed here:", version("numpy"))
````

**What just happened?** In a venv, `sys.prefix` (where *this* environment lives) differs from `sys.base_prefix` (the Python it was created from), so the check prints `True` when you use a venv as recommended. (With conda environments it can print `False`; that's fine.) We also asked which `numpy` version is installed in this environment.

### What is a requirements file?

A **requirements file** (usually `requirements.txt`) is a plain-text shopping list of packages, one per line, often with versions:

| Line | Meaning |
|---|---|
| `numpy==2.1.3` | exactly this version (**pinned**: most reproducible) |
| `pandas>=2.2` | this version or newer |
| `scikit-learn` | any version |
| `# a comment` | ignored |

Let's write a tiny one into our temp folder and read it back like pip would.

````python
req_file = workdir / "requirements.txt"
req_file.write_text("# my project's packages\nnumpy==2.1.3\npandas>=2.2\nscikit-learn\n")

for line in req_file.read_text().splitlines():
    line = line.strip()
    if not line or line.startswith("#"):          # skip blank lines and comments
        continue
    print("pip would install ->", line)
````

**What just happened?** We created a requirements file with one comment and three packages. Running `python -m pip install -r requirements.txt` in a terminal would install all three. Sharing this file (instead of saying "install numpy and some other stuff") lets anyone rebuild the same environment.

## 5. Reading error messages

Errors are normal. Professionals see dozens a day. The skill is **reading** them. When Python hits a problem it stops and prints a **traceback**: a report of where it was and what went wrong.

### How to read a traceback: start at the BOTTOM

```
Traceback (most recent call last):            <- "here are the steps that led here" (oldest first)
  File "/tmp/ipykernel_1234/567.py", line 3, in <module>     <- 2. WHERE: file and line number
    total = price * quantity                   <-    the line that failed
                    ^^^^^^^^                   <- 3. the carets point at the exact part
NameError: name 'quantity' is not defined      <- 1. READ THIS FIRST -> error TYPE: message
```

1. **The last line** is the most important: the **error type** (e.g. `NameError`) and a **message** in plain English.
2. Just above it: the **file and line number** where it happened, and the line of code itself.
3. The **arrow**: newer Python versions (3.11+) draw `^^^^` carets under the exact part that failed. Jupyter's own red error display uses an arrow like `----> 3` to point at the line.
4. If the traceback is long (errors inside libraries), look for the **lowest frame that is your own code**: that's usually where the fix goes.
5. In Jupyter, a file name like `/tmp/ipykernel_.../....py` just means "a notebook cell".

Below we trigger each common error **on purpose**, catch it, and print its traceback with `traceback.print_exc()`, which prints the same report Python would show, without stopping the notebook.

````python
import re                        # text search-and-replace (used to shorten long paths)
import traceback                 # formats tracebacks of errors we catch
import numpy as np               # used in the last error example

def show(err_type):
    "Print the traceback of the error being handled right now (like traceback.print_exc())."
    report = traceback.format_exc()                                  # the same text print_exc() prints
    report = re.sub(r'File ".*?site-packages', 'File ".../site-packages', report)  # shorten library paths
    print(report)
    print(f"-> Caught: {err_type.__name__}. The notebook continues.")
````

**What just happened?** Nothing visible: we only defined a helper called `show` that prints the traceback of the error currently being handled. In your own code you can simply call `traceback.print_exc()` inside an `except` block; our helper prints the same report but shortens long library folder paths (which would otherwise show where Python is installed on this computer).

### 1) `NameError`: a name Python doesn't know

````python
score = 42
try:
    print(scroe)                 # typo: 'scroe' instead of 'score'
except NameError:
    show(NameError)
````

**What just happened?** Last line: `NameError: name 'scroe' is not defined`. Recent Python versions even add a hint: *Did you mean: 'score'?*

**Causes:** a typo, using a variable before creating it, or forgetting to run the cell that defines it (section 1!). Also forgetting quotes: `print(hello)` instead of `print("hello")`.

**Fix:** correct the spelling: `print(score)`, or run the defining cell first.

### 2) `TypeError`: the wrong type of value for an operation

````python
age = 30
try:
    message = "I am " + age + " years old"     # can't glue text (str) and a number (int) with +
except TypeError:
    show(TypeError)
````

**What just happened?** `TypeError: can only concatenate str (not "int") to str`. The **type** of a value is what kind of thing it is: `str` (text), `int` (whole number), `float` (decimal number), `list`, and so on. `+` between text and a number is not allowed.

**Fix:** convert the number to text: `"I am " + str(age) + " years old"`, or better, use an f-string: `f"I am {age} years old"`. Other common TypeErrors: calling a function with the wrong number of arguments, or calling something that is not a function.

### 3) `IndexError`: a list position that doesn't exist

````python
colors = ["red", "green", "blue"]      # positions 0, 1, 2 (Python counts from 0!)
try:
    print(colors[3])                   # there is no position 3
except IndexError:
    show(IndexError)
````

**What just happened?** `IndexError: list index out of range`. A list of 3 items has positions **0, 1, 2**. Position 3 would be a fourth item.

**Fix:** use a valid position: `colors[2]` or `colors[-1]` (the last item). Check the length with `len(colors)`. This error often appears in loops that go one step too far ("off-by-one").

### 4) `KeyError`: a dictionary key that doesn't exist

````python
person = {"name": "Ana", "city": "Lisbon"}    # a dictionary: keys -> values
try:
    print(person["age"])                      # there is no key "age"
except KeyError:
    show(KeyError)
````

**What just happened?** `KeyError: 'age'`. The message is just the missing key. Dictionaries (and pandas column names!) must be spelled exactly, including upper/lower case.

**Fix:** check what keys exist with `person.keys()`, fix the spelling, or use `person.get("age", "unknown")`, which returns a default instead of crashing. In pandas, `df["Age"]` vs `df["age"]` causes the same error; print `df.columns` to check.

### 5) `ValueError`: the right type, but an unacceptable value

````python
user_input = "twelve"
try:
    number = int(user_input)                  # int() needs digits, like "12"
except ValueError:
    show(ValueError)
````

**What just happened?** `ValueError: invalid literal for int() with base 10: 'twelve'`. `int()` accepts text (the type is fine), but only text that looks like a whole number (the value is not).

**Fix:** clean or validate the input first: `int("12")` works; `"12.5"` needs `float()`. A `ValueError` from a library (e.g. scikit-learn) often means "your data has the wrong shape or contains missing values", so read the message carefully.

### 6) `ZeroDivisionError`: dividing by zero

````python
total_points, games_played = 120, 0
try:
    average = total_points / games_played     # division by zero is undefined
except ZeroDivisionError:
    show(ZeroDivisionError)
````

**What just happened?** `ZeroDivisionError: division by zero`. The carets point at the division.

**Fix:** decide what should happen when the count is zero, and check first:
`average = total_points / games_played if games_played > 0 else 0`.

### 7) `ModuleNotFoundError`: a library that isn't installed (or is misspelled)

````python
try:
    import sklearnn                         # typo: the real name is sklearn
except ModuleNotFoundError:
    show(ModuleNotFoundError)
````

**What just happened?** `ModuleNotFoundError: No module named 'sklearnn'`.

**Causes and fixes:**
- A typo: fix the name (`import sklearn`).
- The package isn't installed **in the environment the kernel uses**: install it (`%pip install ...`, see section 2) or select the right kernel (menu **Kernel → Change kernel**).
- The *install* name differs from the *import* name: you `pip install scikit-learn` but `import sklearn`; `pip install opencv-python` but `import cv2`. The library's documentation tells you both.

### 8) `FileNotFoundError`: a file that isn't where you said

````python
try:
    with open("data/missing_file.csv") as f:      # a RELATIVE path, resolved from the kernel's cwd
        contents = f.read()
except FileNotFoundError:
    show(FileNotFoundError)
````

**What just happened?** `FileNotFoundError: [Errno 2] No such file or directory: 'data/missing_file.csv'`. Notice the traceback has **two** steps this time: first *our* cell's line, then a line inside IPython's own code (Jupyter quietly wraps `open` with a safer version, in `.../site-packages/IPython/...`). The bug is never in that library frame; look for the lowest frame that is **your** code. Python looked for `data/missing_file.csv` **relative to the current working directory** (here `01_Python`), and it isn't there.

**Fix:** check where you are (`%pwd` or `Path.cwd()`), check the file really exists and its exact name (`%ls`, file names are case-sensitive on Linux/macOS), and build paths with `pathlib`, e.g. `Path("data") / "file.csv"`.

### 9) Shape mismatch in NumPy

````python
a = np.ones((3, 2))          # 3 rows, 2 columns
b = np.ones((3, 3))          # 3 rows, 3 columns
print("a.shape =", a.shape, "| b.shape =", b.shape)
try:
    c = a + b                # element-by-element addition needs compatible shapes
except ValueError:
    show(ValueError)
````

**What just happened?** `ValueError: operands could not be broadcast together with shapes (3,2) (3,3)`. NumPy adds arrays element by element, so their **shapes** (rows, columns) must match, or follow NumPy's "broadcasting" rules (covered in [`02_Data_Analysis/02_numpy.ipynb`](../02_Data_Analysis/02_numpy.ipynb)). Matrix multiplication with `@` has its own rule (inner sizes must match) and gives a similar error.

**Fix:** **print the shapes** of everything involved (`a.shape`, `b.shape`), and find which array is not what you expected. Shape errors are the most common bug in ML and deep-learning code.

### Summary table

| Error | Usually means | First thing to try |
|---|---|---|
| `NameError` | Unknown name: typo or cell not run | Check spelling; Restart and Run All |
| `TypeError` | Operation doesn't fit the type | `print(type(x))`; convert with `str()`, `int()`, `float()` |
| `IndexError` | List position too big | `print(len(x))`; remember counting starts at 0 |
| `KeyError` | Dictionary key / column name missing | `print(d.keys())` or `print(df.columns)` |
| `ValueError` | Right type, bad value (or bad shape) | Print the value or shape; read the message |
| `ZeroDivisionError` | Divided by zero | Check the divisor first |
| `ModuleNotFoundError` | Not installed, wrong env, or typo | Check the name; install into the kernel's env |
| `FileNotFoundError` | Wrong path or cwd | `Path.cwd()`, `%ls`, use `pathlib` |
| `AttributeError` *(bonus)* | Object has no such method/attribute | `print(type(x))`, `dir(x)` |

## 6. A debugging checklist

**Debugging** = finding and fixing the cause of a problem (a "bug"). Not every bug raises an error: the worst ones just give a wrong answer. Here is a calm, repeatable routine.

1. **Read the error's last line**, then find your line in the traceback (section 5).
2. **Inspect, don't guess**: print the values, their **types** and their **shapes** right before the failing line.
3. **Add `assert` checks**: `assert condition, "message"` crashes immediately with your message if the condition is false. It catches bad data early, close to the cause.
4. **Shrink the input**: try 3 rows instead of 3 million. Small inputs run instantly and you can check them by hand.
5. **Change one thing at a time**, then re-run. (And **Restart and Run All** if in doubt.)
6. **Rubber-duck it**: explain your code line by line, out loud, to someone or something (programmers literally use a rubber duck). Saying "and this line... oh!" finds a surprising number of bugs.
7. **Search the error text**: copy the last line, remove your own variable names and file paths, and search for it. Official docs and Q&A sites usually have the answer.
8. **Make a minimal reproducible example** (below) and ask for help.

Let's practise steps 2-4 on a realistic bug: scores read from a file arrive as **text**, not numbers.

````python
raw_scores = ["90", "85", "77"]           # values read from a CSV file often arrive as TEXT

try:
    average = sum(raw_scores) / len(raw_scores)
except TypeError:
    show(TypeError)

print("Inspect -> value:", raw_scores, "| type of first item:", type(raw_scores[0]).__name__)
````

**What just happened?** `sum` failed with `TypeError: unsupported operand type(s) for +: 'int' and 'str'` (`sum` starts from the number 0 and tries to add text to it). Instead of guessing, we **inspected** the data: the items are `str` (text). Now the fix is obvious: convert them to numbers. And we add an `assert` so this exact problem can never sneak through silently again.

````python
scores = [float(s) for s in raw_scores]                          # convert every item to a number
assert all(isinstance(s, float) for s in scores), "scores must be numbers"   # check our assumption
assert len(scores) > 0, "no scores to average"                   # guard against dividing by zero
print("average:", round(sum(scores) / len(scores), 2))

try:
    labels, features = [0, 1, 1], [[1.0], [2.0]]                 # 3 labels but only 2 rows of features
    assert len(labels) == len(features), f"{len(labels)} labels but {len(features)} feature rows"
except AssertionError:
    show(AssertionError)
````

**What just happened?** After converting, the average is **84.0**. The second `assert` deliberately failed, and its `AssertionError` shows **our own clear message** ("3 labels but 2 feature rows"). An assert costs one line and turns a confusing error later (deep inside a library) into an obvious one right where the problem starts.

### A minimal reproducible example (MRE)

When you ask for help, the most useful thing you can share is the **smallest complete piece of code that still shows the problem**, that anyone can copy, paste and run. Making one often reveals the bug by itself.

```python
# Python 3.12, numpy 2.x
import numpy as np
a = np.ones((3, 2))
b = np.ones((3, 3))
c = a + b   # ValueError: operands could not be broadcast together with shapes (3,2) (3,3)
# Expected: a 3x2 result. Why are the shapes different?
```

### How to ask for help well

| Do | Don't |
|---|---|
| Say what you **expected** and what **actually happened** | Write only "it doesn't work" |
| Paste the **full error text** (as text, not a screenshot) | Paste only the last word of the error |
| Include a **minimal reproducible example** | Paste your whole 500-line notebook |
| Give versions (Python, key libraries) and your operating system | Assume everyone has your setup |
| Say what you **already tried** | Make helpers repeat your steps |
| **Remove secrets, passwords and personal data** first | Paste API keys or private data |

## 7. Git in 5 minutes

**Git** is a **version control** tool: it records snapshots of your project's files over time, so you can see what changed, go back, and work with others without overwriting each other. Think of it as **save points in a video game** for your code.

| Word | Meaning |
|---|---|
| **Repository (repo)** | A project folder that Git is tracking (it keeps its history in a hidden `.git` sub-folder) |
| **Commit** | One saved snapshot, with a message describing the change |
| **Staging** (`git add`) | Choosing which changes go into the next commit |
| **`.gitignore`** | A file listing things Git must **never** track: big data, secrets, environments |
| **Remote** (e.g. GitHub) | A copy of the repo on a server, for backup and sharing (covered in `20_Git_and_DevOps`) |

The core loop is: **edit files → `git add` → `git commit` → repeat**.

We'll practise in a **throwaway repository inside a temporary folder**. To keep it fully separate from any Git settings on your computer, the helper below tells Git to ignore your personal configuration, and we set a pretend identity (`Student`, `student@example.com`) for this demo repo only.

````python
repo = workdir / "my_first_repo"
repo.mkdir()

git_env = dict(os.environ,
               GIT_CONFIG_GLOBAL=os.devnull, GIT_CONFIG_NOSYSTEM="1",    # ignore personal + system Git settings
               GIT_AUTHOR_DATE="2026-01-01T12:00:00", GIT_COMMITTER_DATE="2026-01-01T12:00:00")  # fixed dates

def git(*args):
    "Run `git -C <repo> <args>` (like typing git in a terminal inside the repo) and print the output."
    result = subprocess.run(["git", "-C", str(repo), *args], capture_output=True, text=True, env=git_env)
    output = hide_tmp(result.stdout + result.stderr).strip()
    if output:                                   # many git commands print nothing on success
        print(output)

try:
    git("-c", "init.defaultBranch=main", "init", "-q")   # 1. turn the folder into a Git repository
    git("config", "user.name", "Student")                # 2. a pretend identity for THIS repo only
    git("config", "user.email", "student@example.com")
    git_ok = True
    print("Repository created. Contains a .git folder?", (repo / ".git").is_dir())
except FileNotFoundError:
    git_ok = False
    print("Git is not installed on this computer; skipping the Git demo. See START_HERE.md.")
````

**What just happened?** `git init` turned the empty folder into a repository (it created the hidden `.git` folder). `git config user.name ...` (without `--global`) saved a name and email **only inside this demo repo**; every commit records who made it. On your own machine you would set your real details once with `git config --global user.name "Your Name"`.

Next: create some files, including two that must **never** be committed: a data folder and a `.env` file with a (fake) secret. Then write a `.gitignore`.

````python
if git_ok:
    (repo / "train.py").write_text("print('training...')\n")               # code: should be tracked
    (repo / "data").mkdir()
    (repo / "data" / "big_dataset.csv").write_text("a,b\n1,2\n")          # data: usually NOT in Git
    (repo / ".env").write_text("DEMO_SERVICE_API_KEY=fake-not-a-real-key\n")  # secrets: NEVER in Git
    (repo / ".gitignore").write_text("# never track these\ndata/\n.env\n.venv/\n__pycache__/\n")
    git("status", "--short")      # short status: ?? = new file Git is not tracking yet
````

**What just happened?** `git status --short` lists untracked files with `??`. You should see only **`.gitignore`** and **`train.py`**. The `data/` folder and `.env` are invisible to Git because `.gitignore` lists them. (`.venv/` and `__pycache__/` are listed too, for the future.)

Now stage and commit, then look at the history.

````python
if git_ok:
    git("add", ".")                                   # stage everything not ignored
    git("commit", "-q", "-m", "Add training script and .gitignore")   # save a snapshot with a message
    git("log", "--format=%h  %an <%ae>  %ad  %s", "--date=short")    # history: id, author, date, message
    print("Files tracked by Git:")
    git("ls-files")
````

**What just happened?** `git add .` staged every file that is not ignored, and `git commit -m "..."` saved the snapshot. `git log` shows the commit: a short **commit id** (a code made from the content; yours may look different), the author `Student <student@example.com>`, the date and the message. `git ls-files` confirms only `.gitignore` and `train.py` are tracked: **no data, no secrets**.

Now change a file and see exactly what changed with `git diff`.

````python
if git_ok:
    (repo / "train.py").write_text("print('training...')\nprint('accuracy: 0.91')\n")  # edit the script
    git("diff")                                                     # show line-by-line changes
    git("commit", "-q", "-am", "Print accuracy")                   # -a stages tracked files, -m message
    git("log", "--oneline")                                        # compact history, newest first
````

**What just happened?** `git diff` shows the change: lines starting with **`+`** were added (lines with `-` would be removed). We committed it, and `git log --oneline` now lists **two commits**, newest first. You can always go back to the first one; that's the safety net.

⚠️ **Common mistake:** committing a secret and then deleting it in the next commit. **It is still in the history**, and anyone with the repo can see it. If a real key is ever committed (especially if pushed to GitHub), treat it as leaked: revoke it with the provider and create a new one.

Learn branches, GitHub, pull requests and Git for ML in [`20_Git_and_DevOps/01_git_basics.ipynb`](../20_Git_and_DevOps/01_git_basics.ipynb) and [`20_Git_and_DevOps/02_git_for_ml.ipynb`](../20_Git_and_DevOps/02_git_for_ml.ipynb).

## 8. Secrets: keep API keys out of notebooks

An **API key** is a password that lets a program use an online service (for example a paid LLM API). Anyone who has it can use the service **as you**, often on your bill.

**Golden rule: never type a real key into a notebook or script.** Notebooks get shared, committed to Git and uploaded, and the key goes with them (outputs included).

Instead, keep secrets in **environment variables**: named values that your operating system passes to programs. Python reads them with `os.environ.get("NAME")`.

How to set one (for the current terminal session):

| Shell | Command |
|---|---|
| Linux / macOS (bash, zsh) | `export DEMO_SERVICE_API_KEY="..."` |
| Windows PowerShell | `$env:DEMO_SERVICE_API_KEY = "..."` |
| Windows Command Prompt | `set DEMO_SERVICE_API_KEY=...` |

Then start Jupyter **from that same terminal** so the kernel inherits it.

A common alternative is a **`.env` file**: a small text file of `NAME=value` lines in your project folder, **listed in `.gitignore`**, loaded at startup (often with the optional `python-dotenv` package). Here is the safe reading pattern, with a fake name that is surely not set on your machine:

````python
api_key = os.environ.get("DEMO_SERVICE_API_KEY")       # returns None if it is not set (no crash)

if api_key is None:
    print("DEMO_SERVICE_API_KEY is not set. Set it in your terminal or .env file; never in the notebook.")
else:
    print("Key found:", api_key[:4] + "..." )          # if you must show it, show only a few characters
````

**What just happened?** `os.environ.get(...)` returned `None` because that variable doesn't exist, and the code printed a helpful message instead of crashing. (`os.environ["NAME"]` with square brackets would raise a `KeyError`, which you already know how to read.)

Now let's load the fake `.env` file we put in the demo repo. We use `python-dotenv` if it's installed, otherwise a tiny reader of our own; both do the same thing.

````python
env_file = repo / ".env"                         # the fake secrets file (ignored by Git above)
try:
    from dotenv import dotenv_values             # optional package: python-dotenv
    secrets = dotenv_values(env_file)
except ImportError:                              # not installed? read NAME=value lines ourselves
    secrets = dict(line.split("=", 1) for line in env_file.read_text().splitlines() if "=" in line)

os.environ.update(secrets)                       # make them available as environment variables
key = os.environ.get("DEMO_SERVICE_API_KEY")
print("Loaded a key of length", len(key), "starting with", repr(key[:4] + "..."))
del os.environ["DEMO_SERVICE_API_KEY"]           # tidy up: remove the fake key again
````

**What just happened?** The key came from the `.env` file into the environment, and the code used it without the value ever being typed in the notebook. We printed only its length and first 4 characters, then removed it. Note that `.env` sits next to `.gitignore`, which keeps it out of Git.

⚠️ **Common mistake:** printing a key "just to check it loaded". That value is then saved in the notebook's output and shared with it. Print `key is not None` or its length instead.

💡 **Tip:** all notebooks in this repo run **without** any API key. Where an online service is mentioned, a local model or a mock is used instead.

## Recap

- A **notebook** is cells; the **kernel** runs them and remembers variables **in the order you ran them**. The `[n]` execution count shows that order. When in doubt: **Kernel → Restart and Run All**.
- **Markdown** cells are formatted text; **code** cells are Python. Get help with `help(x)`, `x?` or **Shift + Tab**; save typing with **Tab**.
- **Magic commands** (`%time`, `%timeit`, `%pwd`, `%ls`, `%pip`) are Jupyter-only helpers. Install packages with `%pip install`, not `!pip install`.
- The **terminal** runs commands in a **current working directory**. **Relative** paths depend on it; **absolute** paths don't (but only work on one machine). Run scripts with `python script.py`; exit code 0 = success.
- Use one **virtual environment** per project and record packages in **`requirements.txt`**.
- Read tracebacks **from the bottom**: error type + message, then your line and the `^^^` carets.
- Debug by **inspecting** values, types and shapes, adding **asserts**, shrinking inputs and making a **minimal reproducible example**.
- **Git**: `init` → `add` → `commit` → `log` / `diff`. Use **`.gitignore`** for data, environments and secrets.
- **Secrets** live in environment variables or an ignored `.env` file, read with `os.environ.get(...)`. Never in a notebook.
