# Google Cloud Platform (GCP) for AI/ML

## Overview
GCP is Google's cloud platform, home to Vertex AI, BigQuery, and the TPU hardware that powers Google's own research. It's particularly strong for large-scale data analytics combined with ML.

---

## 1. Vertex AI

Google's unified ML platform (replaced AI Platform):

| Component | Description |
|-----------|-------------|
| Workbench | Managed JupyterLab notebooks |
| Training | Custom training jobs (any framework) |
| Pipelines | Kubeflow-based ML pipelines |
| Model Registry | Version and manage models |
| Feature Store | Online + offline feature serving |
| Prediction | Deploy endpoints (online, batch) |
| Experiments | Track hyperparameters and metrics |
| Model Monitoring | Detect drift in production |

### Vertex AI Training
```python
# Custom training job
from google.cloud import aiplatform

job = aiplatform.CustomTrainingJob(
    display_name='my-training-job',
    script_path='train.py',
    container_uri='us-docker.pkg.dev/vertex-ai/training/pytorch-gpu.1-13:latest',
    requirements=['transformers', 'datasets']
)
model = job.run(replica_count=1, machine_type='n1-standard-8', accelerator_type='NVIDIA_TESLA_T4')
```

## 2. BigQuery

Serverless data warehouse SQL at petabyte scale:
- Pay per query (TB scanned)
- Columnar storage
- Integration with Vertex AI via BigQuery ML

### BigQuery ML
Train ML models directly in SQL:

```sql
-- Train a logistic regression model
CREATE OR REPLACE MODEL `my_dataset.churn_model`
OPTIONS(
  model_type='LOGISTIC_REG',
  input_label_cols=['churned']
) AS
SELECT * FROM `my_dataset.training_data`;

-- Evaluate
SELECT * FROM ML.EVALUATE(MODEL `my_dataset.churn_model`);

-- Predict
SELECT * FROM ML.PREDICT(MODEL `my_dataset.churn_model`,
  (SELECT * FROM `my_dataset.new_customers`));
```

Supported model types: Linear Regression, Logistic Regression, K-Means, Matrix Factorization, XGBoost, DNN, AutoML, LLMs (text-bison, embedding models).

## 3. Cloud Run & Cloud Functions

**Cloud Run**: Serverless containers ideal for ML model serving
- Fully managed, scales to zero
- Up to 32GB RAM, 8 vCPUs
- GPU support (NVIDIA L4)
- Deploy any Docker container

**Cloud Functions**: Event-driven functions (like AWS Lambda)
- Gen 2 supports up to 16GB RAM
- Good for lightweight inference triggers

---

## 4. Tensor Processing Units (TPUs)

Google's custom AI accelerators:
- Designed for matrix multiplication: $$C = A \times B$$ at massive throughput
- TPU v4: 275 TFLOPS per chip
- TPU v5e: optimized for inference
- Use with JAX, TensorFlow, PyTorch/XLA
- Available as Cloud TPUs (on-demand) or TPU pods (thousands of chips)

---

## 5. Gemini API via Vertex AI

Access Gemini models with enterprise SLAs:
- Gemini 1.5 Pro (1M context window)
- Gemini 1.5 Flash (fast, cheap)
- Gemini Ultra
- Native multimodal (text, image, video, audio)
- Function calling, grounding with Google Search
