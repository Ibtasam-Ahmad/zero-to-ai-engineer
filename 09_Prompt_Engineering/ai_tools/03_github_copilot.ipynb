# AI Tools GitHub Copilot

## What is GitHub Copilot?

GitHub Copilot is an **AI pair programmer** built into VS Code, JetBrains, Neovim, and GitHub itself. It uses OpenAI Codex/GPT-4o to suggest code completions, answer questions, and write entire functions.

## Core Features

### Inline Completions
- As you type, Copilot suggests completions in **grey ghost text**
- `Tab` to accept, `Esc` to dismiss
- `Alt+]` / `Alt+[` to cycle through alternatives
- Works in all languages: Python, JS, TypeScript, Go, Rust, etc.

### Copilot Chat
Open with `Ctrl+Shift+I` or click the chat icon:
- Context-aware knows your open files
- Persistent conversation
- Supports `@workspace`, `@vscode`, `@terminal` agents

---

## Slash Commands in Copilot Chat

| Command | What it does |
|---------|-------------|
| `/explain` | Explain selected code in plain English |
| `/fix` | Identify and fix bugs in selection |
| `/tests` | Generate unit tests for selected code |
| `/doc` | Generate documentation/docstrings |
| `/optimize` | Suggest performance improvements |
| `/simplify` | Simplify complex code |
| `/new` | Create new file/project scaffold |
| `/newNotebook` | Create Jupyter notebook |

## Custom Instructions

Create `.github/copilot-instructions.md` in your repo:

```markdown
# Copilot Instructions

This is a Python FastAPI project using PostgreSQL.
- Use SQLAlchemy 2.0 async style
- Always include error handling with HTTPException
- Follow our naming convention: snake_case for functions, PascalCase for classes
- Database models are in app/models/, schemas in app/schemas/
```

## Best Practices

| Practice | Detail |
|----------|--------|
| **Write good comments** | Copilot reads your comments to predict intent |
| **Name things clearly** | `calculate_monthly_revenue()` > `calc()` |
| **Keep files focused** | Single-responsibility files = better suggestions |
| **Use type hints** | Better completion accuracy with typed code |
| **Review always** | Never blindly accept security issues can be suggested |
| **Use /explain for onboarding** | Understand existing code quickly |

````python
# ── Copilot Prompt Patterns Writing comments to guide completions ────────────

# Pattern 1: Describe what you want in a comment, then let Copilot complete
# Example of writing code that Copilot would complete well:

# Function: Parse a CSV file and return a list of dicts,
# handling encoding errors and empty rows
def parse_csv(filepath: str, encoding: str = 'utf-8') -> list[dict]:
    # Copilot would generate the implementation here
    import csv
    results = []
    try:
        with open(filepath, encoding=encoding, errors='replace') as f:
            reader = csv.DictReader(f)
            for row in reader:
                if any(row.values()):  # skip empty rows
                    results.append(dict(row))
    except FileNotFoundError:
        raise FileNotFoundError(f"CSV file not found: {filepath}")
    return results

print("parse_csv function defined")
````

````python
# ── Using GitHub API to interact with Copilot programmatically ────────────────
# GitHub Copilot API (for Copilot Business/Enterprise users)

import requests
import os

# Note: Copilot API requires special access. This shows the pattern.
def copilot_chat(message: str, token: str) -> str:
    """Call GitHub Copilot Chat API (requires Copilot license)"""
    headers = {
        "Authorization": f"token {token}",
        "Content-Type": "application/json"
    }
    # This is a conceptual example actual endpoint may differ
    response = requests.post(
        "https://api.github.com/copilot_internal/v2/token",
        headers=headers
    )
    return response.json()

print("Copilot API pattern shown")
print("For actual usage, use VS Code extension or GitHub.com interface")
````

````python
# ── Copilot Chat Slash Command Examples (as prompt strings) ───────────────────

copilot_examples = {
    "/explain": """
        # Select this code in VS Code, then type /explain
        def quicksort(arr):
            if len(arr) <= 1:
                return arr
            pivot = arr[len(arr) // 2]
            left = [x for x in arr if x < pivot]
            middle = [x for x in arr if x == pivot]
            right = [x for x in arr if x > pivot]
            return quicksort(left) + middle + quicksort(right)
        # Copilot explains: recursive divide-and-conquer, pivot selection, etc.
    """,
    "/tests": """
        # Select a function and type /tests
        # Copilot generates:
        def test_quicksort_empty():
            assert quicksort([]) == []
        
        def test_quicksort_single():
            assert quicksort([1]) == [1]
        
        def test_quicksort_sorted():
            assert quicksort([1, 2, 3]) == [1, 2, 3]
        
        def test_quicksort_reverse():
            assert quicksort([3, 2, 1]) == [1, 2, 3]
    """
}

for cmd, example in copilot_examples.items():
    print(f"Command: {cmd}")
    print(example)
    print("-" * 40)
````
