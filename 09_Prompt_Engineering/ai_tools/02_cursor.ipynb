# AI Tools Cursor IDE

## What is Cursor?

Cursor is an **AI-first code editor** built on VS Code. It gives the AI full awareness of your codebase not just the current file enabling more accurate, context-aware assistance.

---

## Core Features

### 1. Tab Completion (Autocomplete)
- Predicts multi-line edits, not just next tokens
- Learns from your recent edits (cursor position awareness)
- Press `Tab` to accept, `Esc` to reject

### 2. Cursor Chat (`Ctrl/Cmd + L`)
Ask questions about your codebase:
- "How does authentication work in this project?"
- "Find all places where we query the database"
- "What does this function do?"

### 3. Inline Edit (`Ctrl/Cmd + K`)
Select code, press `Cmd+K`, type an instruction:
- "Add error handling"
- "Convert to async"
- "Add type hints"
- "Write tests for this function"

### 4. Composer (Multi-file edits)
`Ctrl/Cmd + Shift + I` Describe a feature and Cursor edits multiple files simultaneously.

---

## @-Mention Context System

Bring specific context into any chat or composer:

| @-mention | What it includes |
|-----------|------------------|
| `@file` | Specific file content |
| `@folder` | All files in a folder |
| `@codebase` | Entire indexed codebase |
| `@docs` | Documentation you've added |
| `@web` | Live web search results |
| `@git` | Git history/diffs |
| `@terminal` | Terminal output |
| `@definitions` | Symbol definitions |

---

## Rules for AI (`.cursorrules` / Project Rules)

Place a `.cursorrules` file at project root to set permanent instructions:

```
# .cursorrules
You are an expert Python developer working on a FastAPI project.

Rules:
- Always use type hints
- Follow PEP 8
- Prefer async functions
- Use Pydantic for data validation
- Write docstrings in Google style
- Tests use pytest with fixtures
- Never use print(), use logging instead
```

---

## Agent Mode

Cursor's most autonomous mode:
- Runs terminal commands
- Creates/deletes files
- Reads error output and self-corrects
- Can complete entire features end-to-end

Enable in Composer with the Agent toggle.

---

## Model Selection

Available models in Cursor:
- **claude-sonnet-4-6** (default, recommended for most tasks)
- **claude-opus-4** (complex reasoning)
- **GPT-4o**
- **o1, o3-mini** (for hard reasoning)
- **cursor-small** (fast, for autocomplete)

---

## Privacy Mode

Enable in Settings → Privacy:
- Code is **not stored** on Cursor's servers
- Not used for training
- Suitable for proprietary codebases

---

## MCP in Cursor

Cursor supports MCP (Model Context Protocol) servers:
```json
// .cursor/mcp.json
{
  "mcpServers": {
    "filesystem": {
      "command": "npx",
      "args": ["-y", "@modelcontextprotocol/server-filesystem", "/path/to/project"]
    }
  }
}
```

## Best Practices

| Practice | Why |
|----------|-----|
| Be specific in instructions | "Add input validation" beats "improve this" |
| Use @-mentions liberally | More context = better results |
| Review before accepting | Always read diffs before applying |
| Set up .cursorrules | Removes repetitive instructions |
| Use Composer for features | Chat for questions, Composer for changes |
| Iterate in small steps | Don't ask for everything at once |

````python
# ── Example: Generate .cursorrules for an ML project ─────────────────────────
cursorrules_template = """
# .cursorrules ML Project

You are an expert ML engineer working on a PyTorch deep learning project.

## Tech Stack
- Python 3.11+
- PyTorch 2.x
- Hugging Face Transformers
- FastAPI for serving
- MLflow for experiment tracking
- Pytest for testing

## Code Style
- Always add type hints
- Use dataclasses or Pydantic for configs
- Log with Python logging, never print()
- Use pathlib.Path instead of os.path
- Prefer torch.no_grad() context manager for inference

## Model Code
- Define models as nn.Module subclasses
- Add forward() type hints
- Use device-agnostic code (avoid hardcoding 'cuda')

## Testing
- Use pytest fixtures for model/data setup
- Mock external API calls
- Test shapes explicitly: assert output.shape == expected_shape
"""

with open(".cursorrules_example", "w") as f:
    f.write(cursorrules_template)

print("Generated .cursorrules template")
print(cursorrules_template)
````

````python
# ── Effective Prompt Patterns for Cursor Chat ─────────────────────────────────

cursor_prompt_patterns = {
    "Code Understanding": [
        "@file Explain how this module works, focusing on the data flow",
        "What are the key abstractions in @codebase?",
        "@file Find all the places where side effects occur",
    ],
    "Code Generation": [
        "Add a rate limiter to @file using the token bucket algorithm",
        "Create a pytest fixture for @definitions DatabaseConnection that uses an in-memory SQLite",
        "Generate OpenAPI docs for all endpoints in @folder",
    ],
    "Debugging": [
        "@terminal This error appeared. Find the root cause in @codebase",
        "@file The function returns wrong results for edge cases. Find the bug.",
        "Why is this @git commit causing performance regression?",
    ],
    "Refactoring": [
        "Refactor @file to use dependency injection",
        "Extract the business logic from @file into a separate service class",
        "Convert @file from synchronous to async",
    ]
}

for category, prompts in cursor_prompt_patterns.items():
    print(f"\n{'='*50}")
    print(f"  {category}")
    print(f"{'='*50}")
    for p in prompts:
        print(f"  • {p}")
````
