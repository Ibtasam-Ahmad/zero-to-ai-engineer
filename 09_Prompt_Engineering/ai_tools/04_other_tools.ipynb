````python
# ── Perplexity API Example ────────────────────────────────────────────────────
import os, requests

def perplexity_search(query: str) -> str:
    """Search with Perplexity API returns answer with citations"""
    url = "https://api.perplexity.ai/chat/completions"
    headers = {
        "Authorization": f"Bearer {os.getenv('PERPLEXITY_API_KEY')}",
        "Content-Type": "application/json"
    }
    payload = {
        "model": "sonar-pro",
        "messages": [
            {"role": "system", "content": "Be precise and concise."},
            {"role": "user", "content": query}
        ],
        "return_citations": True,
        "search_recency_filter": "month"
    }
    response = requests.post(url, json=payload, headers=headers)
    result = response.json()
    answer = result['choices'][0]['message']['content']
    citations = result.get('citations', [])
    return answer, citations

# Example
# answer, sources = perplexity_search("What are the latest LLM benchmarks?")
# print(answer)
# print("Sources:", sources)
print("Perplexity API pattern ready set PERPLEXITY_API_KEY to use")
````

````python
# ── Tool Selection Decision Tree ──────────────────────────────────────────────
def recommend_tool(task: str) -> str:
    task = task.lower()
    
    if any(w in task for w in ['research', 'fact', 'latest', 'current', 'news']):
        return "📡 Perplexity AI real-time search with citations"
    elif any(w in task for w in ['ui', 'component', 'react', 'frontend', 'design']):
        return "🎨 v0.dev UI component generation"
    elif any(w in task for w in ['full-stack', 'app', 'deploy', 'prototype', 'quick']):
        return "⚡ Bolt.new full-stack app generation"
    elif any(w in task for w in ['document', 'pdf', 'paper', 'study', 'notes']):
        return "📚 NotebookLM document-grounded AI"
    elif any(w in task for w in ['code', 'debug', 'refactor', 'function', 'codebase']):
        return "💻 Cursor or GitHub Copilot code-aware IDE AI"
    elif any(w in task for w in ['reason', 'math', 'complex', 'hard', 'difficult']):
        return "🧠 Claude Opus or GPT o1 extended reasoning"
    elif any(w in task for w in ['image', 'generate', 'picture', 'visual']):
        return "🖼️  GPT-4o / DALL-E 3 multimodal generation"
    else:
        return "💬 ChatGPT or Claude general purpose assistant"

tasks = [
    "I need to find the latest papers on RAG",
    "Build a login form component in React",
    "Create a full-stack todo app fast",
    "Study my ML textbook PDF",
    "Debug this Python function",
    "Solve a complex math proof"
]

for task in tasks:
    print(f"Task: {task}")
    print(f"Tool: {recommend_tool(task)}")
    print()
````
